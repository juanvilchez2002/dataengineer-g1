```python
# Crear una bd Tienda
tienda_db = client['tienda']
```

```python
# Crear dos colecciones: clientes y pedidos

clientes = tienda_db['clientes']
pedidos = tienda_db['pedidos']
```

```python
# Insertando datos en el documento cliente
clientes.insert_many([
  { "_id": 1, "nombre": "Luis Ramos", "ciudad": "Bogotá" },
  { "_id": 2, "nombre": "Ana Torres", "ciudad": "Lima" },
  { "_id": 3, "nombre": "Maria Paz", "ciudad": "Madrid" }
])
```

```python
# Insertando datos en el documento pedidos
pedidos.insert_many([
    { "_id": 101, "cliente_id": 1, "total": 75 },
    { "_id": 102, "cliente_id": 2, "total": 80 },
    { "_id": 103, "cliente_id": 3, "total": 90 },
    { "_id": 104, "cliente_id": 2, "total": 30 },
    { "_id": 105, "cliente_id": 3, "total": 100 }
])
```

```python
# SELECT 
# p.id, p.total
# c.nombre
# FROM pedidos p
# LEFT JOIN clientes c
# ON p.cliente_id = c.cliente_id

```

```python
resultado = pedidos.aggregate(
    [
        {
            "$lookup": {
                "from": "clientes",
                "localField": "cliente_id",
                "foreignField": "_id",
                "as": "cliente"
            }
        },
        {"$unwind": "$cliente"},
        {"$project": {"total": 1, "nombre": "$cliente.nombre"}}
    ]
)

for r in resultado:
    print(r)
```

```python
resultado = pedidos.aggregate(
    [
        {
            "$lookup": {
                "from": "clientes",
                "localField": "cliente_id",
                "foreignField": "_id",
                "as": "cliente"
            }
        },
        {"$unwind": "$cliente"},
        {"$match": {"total": {"$gt": 50}, "cliente.ciudad": "Lima"}},
        {"$project": {"total": 1, "nombre": "$cliente.nombre"}}
    ]
)

for r in resultado:
    print(r)
```
