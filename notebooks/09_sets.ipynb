### Set o conjuntos


```python
# lista de regiones
regiones = ['Lima', 'Cuzco', 'Lima', 'Arequipa', 'Cuzco', 'Piura']

# conjunto de regiones
set_regiones = set(regiones)

print(f"Lista de regiones: {regiones}")
print(f"Conjunto de regiones (set): {set_regiones}")
```

```python
len(regiones)
```

```python
len(set_regiones)
```

```python
sedes_esperadas = {'Arequipa', 'Piura', 'Lorero'}
```

```python
sedes_regiones = {'Arequipa', 'Cuzco', 'Lima', 'Piura'}
```

### union ( | - or)

```python
sedes_esperadas | sedes_regiones
```

```python
sedes_esperadas.union(sedes_regiones)
```

### interseccion (& - and)

```python
sedes_esperadas & sedes_regiones
```

```python
sedes_esperadas.intersection(sedes_regiones)
```

### diferencia ( R - )

```python
sedes_esperadas - sedes_regiones
```

```python
sedes_esperadas.difference(sedes_regiones)
```

```python
sedes_regiones - sedes_esperadas
```

```python
sedes_regiones.difference(sedes_esperadas)
```

### diferencia simetrica
la union de ambos conjuntos - la intersección

```python
sedes_regiones ^ sedes_esperadas
```

```python
sedes_regiones.symmetric_difference(sedes_esperadas)
```

### # subconjuntos -> si un conjunto esta dentro de un conjunto

```python

sedes_regiones.issubset(sedes_esperadas)
```
