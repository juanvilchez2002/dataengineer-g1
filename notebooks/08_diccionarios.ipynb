### Diccionario

```python
venta = {
    'id': 'V-101',
    'region': 'Lima',
    'cantidad': 100,
    'precio': 205.5
}
```

```python
venta['region']
```

```python
venta['precio']
```

```python
venta['region'] = 'Arequipa'
venta
```

```python
venta['descuento']
```

```python
# cuando un elemento no existe en el diccionario, genera un error, para evitar esto podemos usar
# get, que busca el elemento y podemos decirle que si no lo encuentra nos de un valor determinado
venta.get('descuento', 0)
```

```python
venta['moneda'] = 'PEN'
venta
```

```python
venta.get('moneda', "UDS")
```

```python
# eliminar un elemento de un diccionario
del venta['moneda']
```

```python
'moneda' in venta
```

```python
'cantidad' in venta
```

```python
# lista de las llaves de la lista
list(venta.keys())
```

```python
# lista de los valores de la lista
list(venta.values())
```

```python
# lista de los elementos (items) de un diccionario
list(venta.items())
```

### Actualizar y anidar diccionario

```python
# actualizar muchos elementos en una sola linea con update
# venta['cantidad'] = 500
# venta['precio'] = 1000

venta.update({'cantidad':500, 'precio':1000})
venta
```

```python
# creamos un elemento si no existe y le asignamos un valor por defecto
# venta['descuento'] = 500 -> novato

venta.setdefault('descuento', 1000)
venta.setdefault('precio', 5000)
venta
```

```python
# Anidación
pedido = {
    'id': 'P-9',
    'cliente': {
        'nombre': 'Marisol',
        'ciudad': 'Lima',
        'edad': '18'
    },
    'items': [
        {
            'sku': 'A2',
            'precio': 19.5
        } ,
        {
            'sku': 'A2',
            'precio': 20.55
        }
    ]    
}
```

```python
len(pedido)
```

```python
list(pedido.keys())
```

```python
pedido['cliente']
```

```python
pedido['cliente']['nombre']
```

```python
# items tiene una lista de diccionarios
pedido['items']
```

```python
# accediento al primer elemento de la lista del diccionario
pedido['items'][1]
```

```python
# accediento al primer elemento del diccionario que se encuenta en la lista del diccionario
pedido['items'][1]['sku']
```
