###Metodos de limpieza


```python
nombre = " Ana        Torres"
apellido_materno = "Zeballos"

print(f"Número de caractéres: {len(nombre)}")
print(f"Número de caractéres: {len(apellido_materno)}")
```

```python
# strip() elimina espacios de los extremos
print(len(nombre.strip()))
print(nombre.strip())
```

```python
print(nombre.upper())

print(nombre.lower())

print(nombre.title())
```

```python
nombre.replace("Torres", "Llerena")
```

```python
nombre.replace(" ", "").replace("AT", "A T").title()
```

```python
# divide de acuerdo a lo que se desea ()
nombre.split()
```

```python
# join, une caracteres
" ".join(nombre.split())
```
