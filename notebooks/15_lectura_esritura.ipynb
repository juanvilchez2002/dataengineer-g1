```python
path = "../data/notas01.txt"

# aqui escribe en el archivo, lo sobre escribe borrando el contenido anterior
with open(path, "w") as f:
    f.write("Primera fila\n")
    f.write("Segunda fila\n")
    f.write("Tercera fila\n")
    f.write("Cuarta fila\n")
```

```python
path = "../data/notas01.txt"

# aqui escribe en el archivo, pero no sobreescribe borrando el contenido anterior; aumenta lo que se encuentra en el archivo
with open(path, "a") as f:
    f.write("Primera fila\n")
    f.write("Segunda fila\n")
    f.write("Tercera fila\n")
    f.write("Cuarta fila")
    f.write("Quinta fila")
```

```python
path = "../data/notas.txt"

lista = ['Primera fila', 'Segunda fila', 'Tercera fila']
with open(path, "w") as f:
    for l in lista:
        f.write(l)
```

```python
import io

path = "../data/notas.txt"

lista = ['Primera fila', 'Segunda fila', 'Tercera fila']

def write(f: io.TextIOWrapper, linea: list[str])->None:
    f.writelines([l + "\n" for l in lista])

with open(path, "w") as f:
    write(f, lista)
```

### LECTURA ESCRITURA

```python
from datetime import datetime #fechas y horas
import time #tiempo

# reitentos
total_intentos = 3
n_intento = 0

# lista de errores
errores = []

# bucle de reitentos
while n_intento < total_intentos:
    
    # la ruta del archivo
    path = "../data/notas20260926.txt"

    try:
        # le indicamos la ruta y le indicamos 
        with open(path, "r") as f:
            # descargamos el contenido en una variable
            contenido = f.read()
            print(f"El contenido es: {contenido}")
            break
    except FileNotFoundError as e:
        errores.append(f"{datetime.now()} | Error:{e}| Número de intento: {n_intento + 1}")
    
    n_intento+=1
    time.sleep(10) # le damos un espacio de 10 segundos para que vuelva a intentar de nuevo

for error in errores:
    print(error)
```
