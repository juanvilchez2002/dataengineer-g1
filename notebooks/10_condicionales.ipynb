```python
color = 'verde'

if color == 'verde':
    premio = 10
    print(premio)
elif color == 'azul':
    premio = 5
    print(premio)
else:
    premio = 0
    print(premio)
    
```

```python
registros = []

print("Llenar registros de otro dia...")

registros.extend([1,2,3,4,5])

if registros:
    print("Procesando...")
else:
    print("Los registros están vacioes...")
```

```python
### MAS CODIGO LIMPIO
cantidad_registros = len(registros)
 
if cantidad_registros > 0:
     print("Procesando...")
```
