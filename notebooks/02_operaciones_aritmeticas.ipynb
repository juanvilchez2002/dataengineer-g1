```python
# operación suma
numero_1 + numero_2
```

```python
# operación resta
numero_2 - numero_1
```

```python
# operación división
numero_1 / numero_2
```

```python
# operación división entera
numero_1 // numero_2
```

```python
# operación residuo
numero_2 % numero_1
```

```python
# operación potencia
numero_1 ** numero_2
```
