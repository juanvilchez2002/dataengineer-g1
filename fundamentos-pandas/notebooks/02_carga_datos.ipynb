### LEYENDO UNA APO

```python
import pandas as pd
import requests

# https://open-meteo.com/en/docs

respuesta = requests.get(
    "https://api.open-meteo.com/v1/forecast",
    params={
        "latitude": -16.40,
        "longitude": -71.54,
        "hourly": "temperature_2m",
        "timezone": 'America/Lima',
        "forecast_days":7,
    },
    timeout=30
    )

respuesta.raise_for_status()
```

```python
respuesta.json()
```

```python
import requests
import pandas as pd

url = "https://pokeapi.co/api/v2/pokemon"

respuesta = requests.get(url, params={"limit":20}, timeout=30)

respuesta.raise_for_status()

datos = respuesta.json()

print(datos)

df = pd.DataFrame(datos["results"])

#respuesta.json()
print(df)
```

```python
respuesta = requests.get(url+"/1", timeout=10)

respuesta.raise_for_status()
respuesta.json
```

```python
filas = []

for url_pokemon in df["url"]:
    respuesta = requests.get(url_pokemon, timeout=30)
    respuesta.raise_for_status()
    pokemon = respuesta.json()
    
    filas.append({
        "id": pokemon["id"]
    })
    
df_pokemon = pd.DataFrame(filas)
df_pokemon.head()
```
