### LOC

```python
# indexar ILOC[Filas, Columnas] 
# localia las filas por indice, dado que las columnas no tienen indice se debe usar su etiqueta o nombre

ventas_df.loc[0:4] #trae las filas del index 0 al 4
```

```python
ventas_df.loc[0:4, ['vendedor', 'producto', 'categoria']]
```

```python
# desde el indice 1000 hacia adelante
ventas_df.loc[1000:, ['vendedor', 'producto', 'categoria']]
```

```python
# las primeras 1000
ventas_df.loc[1000:, ['vendedor', 'producto', 'categoria']]
```

ILOC

```python
# a diferencia del loc, te trae en función al integer ejemplo iloc[0:4], te trae 0 al 3, cuatro elementos al igual que las listas osea por posición

ventas_df.iloc[0:4]
```

```python
ventas_df.iloc[0:4, 0:2]
```
