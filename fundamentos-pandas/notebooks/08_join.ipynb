### CONCATENADO DE TABLAS

```python
import pandas as pd

enero = pd.DataFrame({
    'mes': ['Enero'],
    'numero_mes': [1]
})

febrero = pd.DataFrame({
    'mes': ['febrero'],
    'numero_mes': [2]
})

marzo = pd.DataFrame({
    'mes': ['marzo'],
    'numero_mes': [3]
})

print(enero)
print(febrero)
print(marzo)
```

```python
# unir tablas sin criterio, copiar una tras otra
# concatenando por columnas
pd.concat(
    [enero, febrero, marzo],
    axis=1 # concatenar x el eje 1 (eje de las columnas)
)
```

```python
# unir tablas sin criterio, copiar una tras otra
# concatenando por filas
pd.concat(
    [enero, febrero, marzo],
    axis=0 # concatenar x el eje 0 (eje de las filas)
)
```

```python
# unir tablas sin criterio, copiar una tras otra
# concatenando por filas y resetear indice
pd.concat(
    [enero, febrero, marzo],
    axis=0 # concatenar x el eje 0 (eje de las filas)
).reset_index(drop=True) #si no tiene drop=True, te trae el indice como columna
```

```python
# unir tablas sin criterio, copiar una tras otra
# concatenando por filas y resetear indice, traendo el indice como columna
pd.concat(
    [enero, febrero, marzo],
    axis=0 # concatenar x el eje 0 (eje de las filas)
).reset_index() 
```
