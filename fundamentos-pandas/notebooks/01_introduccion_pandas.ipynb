### PANDAS

```python
import pandas as pd

print(pd.__version__)
```

### SERIES Y DATAFRAME

```python
# una serie es un arran 1D
serie = pd.Series(
    [10, 20, 30],
    index = ['a', 'b', 'c'],
    name = 'ventas'    
)

serie
```

```python
# un DataFrame es un array 2D
df = pd.DataFrame({
    "producto": ['A', 'B'],
    'unidades': [10, 20]
    }
)

df
```

### CREACIÓN DE UNA SERIE

```python
serie_ventas = pd.Series(
    [120, 85, 200, 45],
    index = ['enero', 'febrero', 'marzo', 'abril'],
    name = 'ventas'
)

serie_ventas
serie_ventas.min()
```

### DATAFRAME
DESDE UN DICCIONARIO

```python
df = pd.DataFrame({
    'producto': ['laptop', 'monitor', 'teclado'],
    'unidades': [3, 7, 12],
    'precio': [1200, 320, 95]    
})

print(df)
```

LISTA DE DICCIONARIO

```python
df2 = pd.DataFrame(
    [
        {'producto': 'laptop', 'unidades': 3, 'precio': 1200},
        {'producto': 'monitor', 'unidades': 7, 'precio': 320}
    ]
)

df2
```

DESDE UNA LISTA DE LISTA

```python
df3 = pd.DataFrame(
    [
        ['laptop', 3, 1200],
        ['monitor', 7, 350]
    ],
    columns=['producto', 'unidiades', 'precio']
)
df3
```

### COLUMNAS CALCULADAS

```python
df = pd.DataFrame({
    'producto': ['laptop', 'monitor', 'teclado'],
    'unidades': [3, 7, 12],
    'precio': [1200, 320, 95]    
})

df['importe'] = df['unidades']*df['precio']

df
```

```python
# un DataFrame es un conjunto de Series
df['unidades'].min()
```
