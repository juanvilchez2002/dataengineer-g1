### ELIMNAR NULOS

```python
filtro_nulos = (ventas_df['vendedor'].isna()) | (ventas_df['descuento'].isna())
ventas_df[~filtro_nulos]
```

```python
# dropna() -> elimina todos los registros que tiene un campo con valor nulo
ventas_sin_nulo_df = ventas_df.dropna()
ventas_sin_nulo_df.shape
```

```python
ventas_sin_nulo_df.info()
```

```python
# elimina los nulos de la columna vendedor
ventas_sin_nulos_vendedor = ventas_df.dropna(subset='vendedor')
ventas_sin_nulos_vendedor.info()
```

```python
ventas_sin_nulos_descuento = ventas_df.dropna(subset='descuento')
ventas_sin_nulos_descuento.info()
```

### IMPUTACION DE VALORES

```python
valor_media = ventas_df['descuento'].mean()
print(valor_media)
```

```python
valor_mediana = ventas_df['descuento'].median()
print(valor_mediana)
```

```python
# fillna(valor) -> insertar en los valores nulos el sgte valor

ventas_df['descuento'] = ventas_df['descuento'].fillna(valor_mediana)
```

```python
ventas_df.info()
```

```python
# tener cuidado, aqui como es string puede devolver una serie
valor_vendedor_moda = ventas_df['vendedor'].mode().iloc[0]
print(valor_vendedor_moda)
```

```python
ventas_df['vendedor'] = ventas_df['vendedor'].fillna(valor_vendedor_moda)
```
