```python
import pandas as pd

ventas_df = pd.read_csv('../data/ventas.csv')

ventas_df
```

```python
ventas_df['importe'] = ventas_df['unidades'] * ventas_df['precio_unitario']
ventas_df
```

## REPORTE DE LAS VENTAS TODALES POR REGION

```python
# para una sola region
ventas_df[ventas_df['region'] == 'Sur']['importe'].sum()
```

```python
# usamos el groupby(campo a agrupar)
# mas rapido
ventas_df.groupby('region')['importe'].sum()
```

```python
# el promedio por region
ventas_df.groupby('region')['importe'].mean()
```

```python
# varias estadisticas
# para eso se usa .agg() -> se le puede pasar varias estadisticas como lista
ventas_df.groupby('region')['importe'].agg(['sum', 'mean', 'median', 'count'])
```

```python
# varias estadisticas
# le adicionamos un alias
ventas_df.groupby('region')['importe'].agg(
    importe_total = 'sum',
    importe_promedio = 'mean',
    importe_medinana = 'median',
    cantidad_registros = 'count'
)
```

### AGRUPAR X N-REGISTROS

```python
ventas_df.groupby(['region', 'categoria'])['importe'].agg(
    importe_total = 'sum',
    importe_promedio = 'mean',
    importe_medinana = 'median'
)
```

```python
# estadisticas de varias columnas

ventas_df.groupby('region').agg({
    'importe': ['sum', 'mean', 'max'],
    'unidades': ['sum'],
    'producto': ['nunique']
})
```

```python

ventas_df.groupby(['region', 'categoria']).agg({
    'importe': ['sum', 'mean', 'max'],
    'unidades': ['sum'],
    'producto': ['nunique']
})
```

```python
ventas_df.groupby(['region', 'categoria', 'vendedor']).agg({
    'importe': ['sum', 'mean', 'max'],
    'unidades': ['sum'],
    'producto': ['nunique']
})
```
