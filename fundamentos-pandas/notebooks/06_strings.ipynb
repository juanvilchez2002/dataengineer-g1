# ENCONTRAR DUPLICADOS

```python
# una mascara boleana, indica True el registro duplicado osea igual en todos los campos 
filtro = df.duplicated()
filtro
```

```python
# drop_duplicates -> eliminar duplicados
clientes_sin_duplicados = df.drop_duplicates()
clientes_sin_duplicados.head(10)
```

```python
# strip -> elimina espacios al inicial y final
# lower -> pasa a minuscula

clientes_sin_duplicados['email'].str.strip().str.lower()
```

```python
clientes_sin_duplicados['email'] = clientes_sin_duplicados['email'].str.strip().str.lower()
clientes_sin_duplicados.head(10)
```
