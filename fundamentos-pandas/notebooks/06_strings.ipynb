```python
import pandas as pd

df = pd.read_csv('../data/clientes_sucio.csv')

df.head(15)
```

```python
df.info()
```

# ENCONTRAR DUPLICADOS

```python
# una mascara boleana, indica True el registro duplicado osea igual en todos los campos 
filtro = df.duplicated()
filtro
```

```python
# drop_duplicates -> eliminar duplicados
clientes_sin_duplicados = df.drop_duplicates()
clientes_sin_duplicados.head(10)
```

```python
# strip -> elimina espacios al inicial y final
# lower -> pasa a minuscula

clientes_sin_duplicados['email'].str.strip().str.lower()
```

```python
clientes_sin_duplicados['email'] = clientes_sin_duplicados['email'].str.strip().str.lower()
clientes_sin_duplicados.head(10)
```

```python
# title() para nombre y ciudad
clientes_sin_duplicados['nombre'] = clientes_sin_duplicados['nombre'].str.title()
clientes_sin_duplicados['ciudad'] = clientes_sin_duplicados['ciudad'].str.title()
clientes_sin_duplicados.head(10)
```

```python
clientes_sin_duplicados['fecha_nac'] = pd.to_datetime(
    clientes_sin_duplicados['fecha_nac'],
    format='mixed'
)

clientes_sin_duplicados
```

```python
# darle un trataminto de una columna de texto para convertirla a formato numerico y pode realizar operaciones
# usando map y casteo a float64

def conversor_apunto_flotante(x: str)-> str:
    if x:
        return str(x).replace(".", "").replace(",", ".")
    return x
```

```python
#clientes_sin_duplicados['gasto_total'].map(
#    lambda x: str(x).replace(".", "").replace(",", ".")
#)

clientes_sin_duplicados['gasto_total'] = clientes_sin_duplicados['gasto_total'].str.replace(".","").str.replace(",", ".").astype(pd.Float64Dtype())
clientes_sin_duplicados
```

```python
# activo
#clientes_sin_duplicados['activo'] = clientes_sin_duplicados['activo'].str.lower().str.replace(
    #{"si":"True", 'no': "False"}
#)
clientes_sin_duplicados['activo'] = clientes_sin_duplicados['activo'].str.lower().str.replace(
    {"si":"1", 'no': "0"}
).astype(int).astype(bool)

clientes_sin_duplicados
```

```python
# quitando los espacios y simbolos
clientes_sin_duplicados['telefono'] = clientes_sin_duplicados['telefono'].str.replace(r"\D","", regex=True)
clientes_sin_duplicados
```

```python
clientes_sin_duplicados.to_csv("../data/clientes_limpios.csv", index=False)
```

```python
# para guardar en parquet se usa la liberia pyarrow
# uv pyarrow

clientes_sin_duplicados.to_parquet(
    "../data/clientes_limpios.parquet.gzip",
    index=False,
    compression="zstd"
)
```
