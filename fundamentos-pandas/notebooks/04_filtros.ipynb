### FILTROS

```python
import pandas as pd

ventas_df = pd.read_csv("../data/ventas.csv")
ventas_df.head(20)
```

```python
# mascara
ventas_df["region"] == 'Sur'
```

```python
#
ventas_df[ventas_df["region"]=="Sur"]
```

```python
#
filtro = ventas_df["region"]=="Sur"
ventas_df[filtro]
```

```python
# subset= , le decimos de un determinado campo
ventas_df[ventas_df["region"]=="Sur"].value_counts(subset="region")
```

```python
# operador Y -> &            
# operador O -> |
filtro = (ventas_df['region']== "Sur") & (ventas_df["precio_unitario"]>90)
ventas_df[filtro] 
```

```python
ventas_df[filtro]["precio_unitario"].min()
```

```python
ventas_df[filtro]["region"].unique()
```

```python
filtro = (ventas_df['region']=='Sur') | (ventas_df['region']=='Norte')

ventas_df[filtro]
```

```python
ventas_df[filtro]['region'].unique()
```

```python
filtro = ventas_df['descuento'] == 0.1
ventas_df[filtro]
```

```python
ventas_df[filtro]['descuento'].unique()
```

```python
filtro = ventas_df['region']!= 'Sur'

ventas_df[filtro]
```

```python
ventas_df[filtro]['region'].unique()
```

```python
# filtro = (ventas_df['region']=='Sur') | (ventas_df['region']=='Norte')
filtro = ventas_df['region'].isin(['Sur', 'Norte'])

ventas_df[filtro]['region'].unique()
```

```python
filtro = (ventas_df['precio_unitario'] >= 90) & (ventas_df['precio_unitario'] <= 200)
ventas_df[filtro]['precio_unitario'].value_counts() 
```

```python
# equivalente filtro = (ventas_df['precio_unitario'] >= 90) & (ventas_df['precio_unitario'] <= 200)
# cuando el operador es mayor o igual - menor o igual
filtro = ventas_df['precio_unitario'].between(90, 200)
ventas_df[filtro]
```

```python
ventas_df[filtro]["precio_unitario"].value_counts()
```

```python
# filtro
filtro = (ventas_df['categoria'] == "Tecnologia") & (ventas_df['unidades'] > 10)
ventas_df[filtro]
```

```python
# tener en cuenta que el 2do corteche cuando es mas de un campo se pasa como lista
ventas_df[filtro][['categoria', 'unidades']].value_counts()
```
