### FUNCIONES

```python
def sumar(numero1: float, numero2: float) -> float:
    return numero1+numero2

sumar(20, 30)
```

```python
PI = 3.1416

def calc_area_circulo(radio: float, n_decimales: int = 2) -> float:
    return round(PI*radio**2, n_decimales)

calc_area_circulo(3)
```

```python
from datetime import datetime

def imprimir_fecha() -> None:
    print(datetime.now())
    
fecha = imprimir_fecha()

print(fecha)
```
