### Programación Orientada a Objetos (POO)

```python
class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        
    def ladrar(self):
        return f"{self.nombre}  dice: guauu"
    
    def cumplir_anios(self):
        self.edad +=1
```

```python
fido = Perro('Fido', 5)

print(fido.ladrar())
print(fido.edad)

fido.cumplir_anios()
print(fido.edad)
```

```python
toby = Perro('Toby', 10)

toby.ladrar()
```

```python
class Rectangulo:
    def __init__(self, base:float, altura: float) -> None:
        self.base = base
        self.altura = altura
        
    def area(self) -> float:
        return self.altura * self.base
    
    def perimetro(self) -> float:
        return self.altura*2 + self.base*2
```

```python
rectangulo = Rectangulo(10, 3)

print(f"Area: {rectangulo.area()}")
print(f"Area: {rectangulo.perimetro()}")
```

### Encapsulamiento

```python
class CuentaBancaria:
    def __init__(self, titular: str, saldo: float = 0.00) -> None:
        self.titular = titular # atributo publico
        self._saldo = saldo #este atributo debe ser usado desde la clase y no por fuera
        
    def depositar(self, monto: float):
        if self._validar_monto(monto):
            self._saldo += monto
    
    # metodo privado, no necesita ser expuesto        
    def _validar_monto(self, monto: float) -> bool:
        return monto >0
    
    def ver_saldo(self)-> None:
        return self._saldo
    
    def retirar(self, monto: float) ->None:
        if self._validar_monto(monto):
            self._saldo -= monto
```

```python
cuenta_1 = CuentaBancaria("Juan", 100)
print(f"Saldo inicial: {cuenta_1.ver_saldo()}")

cuenta_1.depositar(50)
print(f"Saldo luego de depositar: {cuenta_1.ver_saldo()}")

cuenta_1.retirar(150)
print(f"Saldo luego de retirar: {cuenta_1.ver_saldo()}")
```

### Herencia

```python
class Animal:
    def __init__(self, nombre: str):
        self.nombre = nombre
    
    def presentarse(self)->str:
        return f"Soy {self.nombre}"
    
#mala practica   
class Perro(Animal):
    def ladrar(self)->str:
        return "Guauu!!!"
    
# buena practica, toda clase debe de tener su propio constructor
class Perro(Animal):
    def __init__(sefl, nombre:str)-> None:
        # super() 
        super().__init__(nombre)
        
    def ladrar(self)->str:
        return "Guauuu!!!"
    
class Gato(Animal):
    def __init__(sefl, nombre:str, color: str)-> None:
        # super() 
        super().__init__(nombre)
        self.color = color
        
    def ladrar(self)->str:
        return "Guauuu!!!"
        
```

```python
toby = Perro('toby')
print(toby.ladrar())
print(toby.presentarse())
```

```python
lucifer = Gato('lucifer', 'negro')
lucifer.presentarse()
```

### POLIMORFISMO

```python
class Animal:
    def __init__(self, nombre: str):
        self.nombre = nombre
    
    def presentarse(self)->str:
        return f"Soy {self.nombre}"
    
    def hablar(self) -> str:
        return "Estoy hablando..."
    
class Perro(Animal):
    def __init__(self, nombre: str)-> None:
        super().__init__(nombre)
        
    def hablar(self):
        return "Guauuu!!!"
    
class Gato(Animal):
    def __init__(self, nombre: str, color: str)-> None:
        super().__init__(nombre)
        self.color = color
        
    def hablar(self):
        return "Miauu!!!" 
    
    
    
```

```python
animales = [Perro('Toby'), Gato('Lucifer', 'negro'), Animal('Random')]

for animal in animales:
    print(animal.hablar())
```

### COMPOSICIÓN

HERENCIA -> "ES UN"
COMPOSICIÓN -> "TIENE UN"

```python
class Motor:
    def __init__(self, caballos):
        self.caballos = caballos
        
    def encender(self):
        return "Motor en marcha"

class Coche:
    def __init__(self, marca, caballos):
        self.marca = marca
        self.motor = Motor(caballos) #composición

        
coche = Coche('Kid', 120)
print(coche.motor.encender())   
print(coche.motor.caballos)
print(coche.marca)
```
