```python
precios = [2300, 5000, 9000]

con_igv = []

for precio in precios:
    con_igv.append(round(precio*1.18, 2))
    
con_igv
```

### LISTA POR COMPRESION

```python
con_igv = [round(precio*1.18,2) for precio in precios]

con_igv
```

```python
ventas = [
    {'region': 'Arequipa', 'monto': 6400},
    {'region': 'Puno', 'monto': 0.0},
    {'region': 'Lima', 'monto': 5.0},
    {'region': 'Tacna', 'monto': 0.0},
]

validos = []

for venta in ventas:
    if venta['monto']>0:
        validos.append(venta)

validos
```

```python
validos_1 = [
    venta
    for venta in ventas
    if venta['monto']>0
]

validos_1
```
