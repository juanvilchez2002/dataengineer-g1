### EXCEPCIONES

```python
fila = {
    'id': 'V101',
    'precio': '200.5',
    'cantidad': '100.3'
}

fila
```

```python
try:
    cantidad = int(fila['cantidad'])
except ValueError:
    print("Cantidad no numerica")
    cantidad = 0
else:
    print("La conversión salió bien")
    print(cantidad)
finally:
    print("Siempre ejecuto este bloque")
```

```python
try:
    precio_total = fila['precio'] +50
except TypeError, ValueError:
    precio_total = float(fila['precio']) +50
    print(f"cantidad total {precio_total}")
    
```
