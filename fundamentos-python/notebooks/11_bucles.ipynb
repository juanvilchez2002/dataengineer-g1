### Bucles FOR
para recorrer colecciones, tuplas, listas

```python
regiones = ["Lima]", "Arequipa", "Trujillo", "Tacna"]

for region in regiones:
    print(f"Esta es la región: {region}")
```

```python
ciudad = 'Arequipa'

for letra in ciudad:
    print(letra)
```

```python
venta = {
    'id': 'V-101',
    'region': 'Lima',
    'cantidad': 100,
    'precio': 205.5
}

for x in venta:
    print(x)
```

```python
for key, values in venta.items():
    print(f"La llave es {key} y el valor es {values}")
```

```python
filas = ['V-100;Lima;2', 'V-101;Arequipa;3', 'V-102; Cuzco;4']

for numero_fila, fila in enumerate(filas):
    print(f"La fila: {numero_fila} tiene el valor: {fila}")
```

```python
columnas = ['id', 'region', 'cantidad']
valores = ['V-101', 'Lima', '2']
```

```python
for columna, valor in zip(columnas, valores):
    print(f"{columna} su valor es {valor}")
```

### Bucle WHILE
para condicionales (romper condiciones, reintentos)

```python
intentos = 0

while intentos < 3:
    print(f"Intento: {intentos}")
    intentos +=1
```

```python
numero = int(input("Ingrese número:"))

while numero<10:
    print(f"El número es: {numero}")
    numero = int(input("Ingrese número:"))

print(f"El número ganador es: {numero}")
```

### Control o flujo de ejecución

```python
for numero in range(20):
    if numero % 2 ==0:
        continue
    
    print(f"Número impar: {numero}")
```

```python
for numero in range(20):
    if numero % 2 ==0:
        continue
    
    if numero >= 10:
        break
    
    print(f"Número impar: {numero}")
    
print(f"El bucle finalizo")
```

```python
ciudades = ['Arequipa', 'Puno', 'Tacna', 'Ayacucho', 'Cuzco']
buscar = 'Tacna'

for ciudad in ciudades:
    print(f'Ciudad: {ciudad}')
    if ciudad == buscar:
        print(f'La ciudad heroica')
        break
```
