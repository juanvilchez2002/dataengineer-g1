### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí
inventario_total = {}

for inventario in [inventario_norte, inventario_sur, inventario_centro]:
    for articulo in inventario:
        inventario_total.setdefault(articulo, 0)
        inventario_total[articulo] += inventario[articulo]
        
print(inventario_total)

###

totales = {}

articulos = set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys()            
])

inventarios = [inventario_norte, inventario_sur, inventario_centro]

for articulo in articulos:
    totales[articulo] ={
        'cantidad_total': sum([inventario.get(articulo, 0) for inventario in inventarios]),
        'sucursales': sum([1 for inventario in inventarios if articulo in inventario])        
    }

for articulo in totales:
    print(f"articulo: {articulo} | {totales[articulo]}")
    print('='*50)
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
# Escribe tu solución aquí
resultado = {}

for empleado in empleados:

    departamento = empleado["departamento"]
    nivel = empleado["nivel"]

    if departamento not in resultado:
        resultado[departamento] = {}

    if nivel not in resultado[departamento]:
        resultado[departamento][nivel] = {
            "nombres": [],
            "cantidad": 0,
            "salarios": []
        }

    resultado[departamento][nivel]["nombres"].append(empleado["nombre"])
    resultado[departamento][nivel]["cantidad"] += 1
    resultado[departamento][nivel]["salarios"].append(empleado["salario"])


for departamento, niveles in resultado.items():
    for nivel, datos in niveles.items():
        datos["salario_promedio"] = sum(datos["salarios"]) / datos["cantidad"]
        del datos["salarios"]


print(resultado)
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
# Escribe tu solución aquí
candidatos = {candidato.lower() for candidato in candidatos}


# Contador
conteo = {}
for candidato in candidatos:
    conteo[candidato] = 0

votos_invalidos = 0

for voto in votos:
    voto = voto.strip().lower()
    if voto in candidatos:
        conteo[voto] += 1
    else:
        votos_invalidos += 1

total_validos = sum(conteo.values())

porcentajes = {}

for candidato, cantidad in conteo.items():
    porcentajes[candidato] = cantidad / total_validos * 100


max_votos = max(conteo.values())
ganadores = []

for candidato, cantidad in conteo.items():
    if cantidad == max_votos:
        ganadores.append(candidato)

print("Conteo:", conteo)
print("Votos inválidos:", votos_invalidos)
print("Porcentajes:", porcentajes)

if len(ganadores) == 1:
    print("Ganador:", ganadores[0])
else:
    print("Hay empate:", ganadores)
```

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(lista: list)->dict:
    if not isinstance(lista, list):
        raise ValueError("El argumento no es una lista")
    
    if not lista or min(lista) < 0:
        raise ValueError("La lista esta vacia o tiene un valor negativo")
        
    resultado = {}
    
    resultado['total vendido'] = sum(lista)
    resultado['promedio vendido'] = resultado['total vendido']  / len(lista)
    resultado['maximo vendido'] = max(lista)
    resultado['minimo vendido'] = min(lista)    
    resultado['ventas mayores promedio'] = sum(1 for listax in lista if listax > resultado['promedio vendido'])   
    resultado['ventas menores 100'] = sum(1 for listax in lista if listax < 100)
    
    return resultado

analizar_ventas([1])
```

### Pregunta 5.2: Validar una contraseña y explicar los errores

Crea una función `validar_password(password)` que determine si una contraseña cumple todas las reglas de seguridad.

**Requisitos:**
- Debe tener al menos 8 caracteres.
- Debe contener una letra mayúscula, una minúscula y un número.
- Debe incluir al menos uno de estos símbolos: `@`, `#`, `$` o `%`.
- No debe contener espacios.
- Debe retornar una tupla con el formato `(es_valida, errores)`, donde `errores` sea una lista.

```python
password_prueba = 'Python123'
```

```python
# Escribe tu solución aquí
def validar_password(password: str)-> tuple:
    if not isinstance(password, str):
        raise ValueError("No es un string")
    
    es_valida = False
    
    errores =[]
    
    is_longitud =  len(password) >= 8
    is_letras = any(passw.isdigit() for passw in password) and any(passw.isupper() for passw in password) and any(passw.islower() for passw in password)
    
    #simbolos = '@' in password or '#' in password or '$' in password or '%' in password
    simbolos = '@#$%'
    is_simbolos = any(caracter in simbolos for caracter in password)
    
    is_espacios = any(caracter.isspace() for caracter  in password)
    
    if not is_longitud:
        es_valida = False
        errores.append('debe al menos tener 8 caracteres')
    
    if not is_letras:
        es_valida = False
        errores.append('Debe contener una letra mayúscula, una minúscula y un número.')
        
    if not is_simbolos:
        es_valida = False
        errores.append('Debe incluir al menos uno de estos símbolos: `@`, `#`, `$` o `%`.')
            
    if is_espacios:
        es_valida = False
        errores.append('No debe contener espacios')
    
    return (es_valida, errores)
    
    
validar_password(password_prueba)
```

- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
# Escribe tu solución aquí
def reservar_asientos(sala, asientos_solicitados):
    total_filas = len(sala)
    total_columnas = len(sala[0])
    
    for fila_u, columna_u in asientos_solicitados:
        fila = fila_u - 1
        columna = columna_u -1        
                
        if fila < 0 or fila>= total_filas:
            return f"La fila {fila} no existe."
        
        if columna <0 or columna >= total_columnas:
            return f"La colmuna {columna} no existe."
        
        estado_asiento = sala[fila][columna]
        
        if estado_asiento =='O':
            return f"El asiento {fila}:{columna} no esta libre."
        
    for fila, columna in asientos_solicitados:
        fila = fila - 1
        columna = columna - 1
        
        sala[fila][columna] = 'O'
        
    return f"Se reservaron los asientos {asientos_solicitados} en la sala {sala}"

reservar_asientos(sala, asientos)
```

### Pregunta 6.1: Clase CuentaBancaria

Crea una clase `CuentaBancaria` con los atributos `titular` y `saldo` (el saldo inicial por defecto será 0).

- Método `depositar(monto)`: suma al saldo, solo si el monto es positivo.
- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# No hay datos de entrada: crea tus propios objetos para probar la clase.
class CuentaBancaria:
    def __init__(self, titular):
        self.titular = titular
        self.saldo = 0
        
    def depositar(self, monto_depositar):
        if monto_depositar <= 0:
            return('El monto a depositar es menor o igual que 0')        
        self.saldo += monto_depositar
            
    def retirar(self, monto_retirar):
        if monto_retirar <= 0:
            return("El monto a retirar es menor o igual que 0")        
        if monto_retirar > self.saldo:
            return("Fondos insuficiente")        
        self.saldo +-monto_retirar
        return f"Saldo disponible: {self.saldo}"

    def consultar_saldo(self):
        return f"{self.titular} su saldo disponible es: S/. {self.saldo}"
```

```python
# Escribe tu solución aquí
juan = CuentaBancaria('Juan')
juan.depositar(150)
print(juan.consultar_saldo())
print(juan.retirar(50))
print(juan.retirar(200))
print(juan.retirar(-10))
```

*Objetivo: composición ("tiene un") frente a herencia ("es un").*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
class Motor:
    def __init__(self, tipo, caballos):
        self.tipo = tipo
        self.caballos = caballos
        self.encendido = False
        
    def arrancar(self):
        if self.encendido:
            return "El motor ya esta encendido"
        
        self.encendido = True
        return "Arranco el motor"

class Vehiculo():
    def __init__(self, marca, motor):
        self.marca = marca
        self.motor = motor

    def arrancar(self):
        return self.motor.arrancar()
```
