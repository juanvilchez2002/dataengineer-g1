CASTEO DE VARIABLES

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))
```

```python
precio = "145.3"
print(type(precio))

precio = float(precio)
print(type(precio))

precio = int(precio)
print(type(precio))
```

```python
activo = 0
print(type(activo))

activo = bool(activo)
print(type(activo))
```

```python
activo = -1
print(activo)
print(type(activo))

activo = bool(activo)
print(activo)
print(type(activo))
```
